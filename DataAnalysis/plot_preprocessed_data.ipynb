```python
# import necessary packages
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np

import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio

import os
```

```python
data_folder = os.path.join('..', 'PreprocessedDataML', 'HookRemovedData')

output_folder = os.path.join(data_folder, 'Figures')
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

for file in os.listdir(data_folder):
    if os.path.isfile(os.path.join(data_folder, file)) == False:
        continue
    
    df = pd.read_csv(os.path.join(data_folder, file))
    grouped = df.groupby(["SubID", "SessionID"])

    fig = go.Figure()
    # Iterate over the grouped data by participant and session
    for i, ((sub_id, session_id), group) in enumerate(grouped, start=1):
        # Iterate over trials for each participant and session, and add scatter traces
        for trial_num, trial_group in group.groupby('TrialID'):
            # (Single Session Figure) Add scatter trace and color bar for right controller
            fig.add_trace(go.Scatter3d(
                x=pd.to_numeric(trial_group['R-HandPosX'], errors='coerce'),
                y=pd.to_numeric(trial_group['R-HandPosY'], errors='coerce'),
                z=pd.to_numeric(trial_group['R-HandPosZ'], errors='coerce'),
                mode='markers',
                marker=dict(
                    size=2,
                    color=trial_group.index,  # set color based on occurrence (use index as a unique identifier)
                    colorscale='viridis',  # colorscale
                    opacity=0.8,
                    colorbar=dict(
                        title='Right',
                        tickvals=[trial_group.index[0], trial_group.index[-1]],
                        ticktext=['Start', 'End'],
                        len=0.6,
                        x=1.05,  # move the color bar to the left (adjust the value as needed)
                        y=.9,
                    )
                ),
                name=f'Sub {int(sub_id)}, Sess {int(session_id)}, Trial {trial_num} - Right Controller',
                showlegend=False
            ))
            # (Single Session Figure) Add scatter trace and color bar for left controller
            fig.add_trace(go.Scatter3d(
                x=pd.to_numeric(trial_group['L-HandPosX'], errors='coerce'),
                y=pd.to_numeric(trial_group['L-HandPosY'], errors='coerce'),
                z=pd.to_numeric(trial_group['L-HandPosZ'], errors='coerce'),
                mode='markers',
                marker=dict(
                    size=2,
                    color=trial_group.index,  # set color based on occurrence (use index as a unique identifier)
                    colorscale='sunset_r',  # colorscale (reverse)
                    opacity=0.8,
                    colorbar=dict(
                        title='Left',
                        tickvals=[trial_group.index[0], trial_group.index[-1]],
                        ticktext=['Start', 'End'],
                        len=0.6,
                        x=.95,  # move the color bar to the left (adjust the value as needed)
                        y=0.9,
                    )
                ),
                name=f'Sub {int(sub_id)}, Sess {int(session_id)}, Trial {trial_num} - Left Controller',
                showlegend=False
            ))
            fig.update_traces(showlegend=True, selector = ({'name': f'Sub {int(sub_id)}, Sess {int(session_id)}, Trial {trial_num} - Left Controller'}))
            fig.update_traces(showlegend=True, selector = ({'name': f'Sub {int(sub_id)}, Sess {int(session_id)}, Trial {trial_num} - Right Controller'}))

        fig.update_layout(
            legend=dict(
                x=.95,
                y=-.5
            ),
            legend_title='Controllers in Use',
            title_text=os.path.basename(file)[:-4],

            scene=dict(
                aspectmode='data'  # Scale axes based on data ranges
            )
        )

    output_path = os.path.join(output_folder, os.path.basename(file[:-4]) + ".html")
    if not os.path.exists(output_path):
        pio.write_html(fig, file=output_path, auto_open=False)
```
